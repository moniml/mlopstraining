```python
import pandas as pd
df=pd.read_csv("/content/social_media_addiction_dataset (1).csv")
df
```

```python
df.info()
```

```python
dummies = pd.get_dummies(df.night_usage, dtype=int)
dummies
```

```python
merged=pd.concat([df,dummies],axis='columns')
merged
```

```python
final=merged.drop(['night_usage'],axis='columns')
final
```

```python
x=final.drop('addiction_level',axis='columns')
y=final['addiction_level']
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict(x_test)
```

Predict addiction level of a student with:
8 hours daily screen time
65 app opens
5 hours sleep
stress level 8
6 study interruptions
night usage = Yes

```python
new=pd.DataFrame([[8,65,5,8,6,0,1]],columns=['daily_screen_time','app_opens','sleep_hours','stress_level','study_interruptions','No','Yes'])
prediction=model.predict(new)
print(prediction)
```
